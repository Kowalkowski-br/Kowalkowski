# 1. ¿Cuántas posibilidades hay sin tener en cuenta las restricciones?

Una solución es un vector $x=(x_1,\dots,x_P)$ donde $x_i\in\{1,\dots,F\}$ es la franja
asignada al partido $i$. Sin ninguna restricción cada partido elige libremente entre las $F$
franjas (se permiten coincidencias), luego el espacio de soluciones tiene tamaño:

$$|S| \;=\; F^{P} \;=\; 10^{10} \;=\; 10\,000\,000\,000$$

# 2. ¿Cuántas posibilidades hay teniendo en cuenta todas las restricciones?

La restricción obliga a que el viernes y el lunes tengan exactamente un partido cada uno:

- Elegir qué partido se juega el **viernes**: $P$ formas.
- Elegir qué partido se juega el **lunes** (distinto del anterior): $P-1$ formas.
- Los $P-2$ partidos restantes se reparten libremente entre las $F-2$ franjas centrales:
  $(F-2)^{P-2}$ formas.

$$\boxed{\;|S_{\text{fact}}| \;=\; P\,(P-1)\,(F-2)^{P-2} \;=\; 10\cdot 9\cdot 8^{8} \;=\; 1\,509\,949\,440\;}$$

Es decir, la restricción elimina uns **85 %** del espacio original.

---

````
from math import comb, factorial

P, F = 10, 10  # partidos, franjas

sin_restr = F ** P                            # 1) sin restricciones
con_restr = P * (P - 1) * (F - 2) ** (P - 2)  # 2) exactamente 1 viernes y 1 lunes


# Vectores de ocupacion a explorar por el algoritmo mejorado (ver preg. 8):
# los P-2 partidos libres se reparten entre las F-2 franjas centrales
ocupaciones = comb((P - 2) + (F - 2) - 1, (F - 2) - 1)

print(f"1) Sin restricciones          F^P              = {sin_restr:,}")
print(f"2) Con restricciones   P(P-1)(F-2)^(P-2)       = {con_restr:,}")
print(f"   -> la restriccion elimina el {100*(1-con_restr/sin_restr):.1f} % del espacio")
print(f"   (ref.) Vectores de ocupacion  C(15,7)        = {ocupaciones:,}")
````

# 3. ¿Cuál es la estructura de datos que mejor se adapta al problema? Argumentación

La estructura que mejor se adapta al problema para explorarlo es el
vector de ocupación:

$$n = (n_1,\dots,n_F), \qquad n_s = \bigl|\{\,i : x_i = s\,\}\bigr|, \qquad \sum_s n_s = P$$

**Argumentación:** Todos los partidos que comparten franja sufren la misma penalización, y esa penalización depende únicamente de cuántos son ($n_s$), no de quiénes son. Por lo tanto la audiencia depende solo de (a) el vector de ocupación $n$ —que fija los pesos de cada franja— y (b) de qué base se empareja con qué peso, esto convierte a $n$ en la representación natural.

Se usan listas indexadas por franja (`COEF`, `PEN`, `ocupacion`,
`pesos`) por su acceso $O(1)$, y un diccionario `frozenset(categorías) → base` para la tabla de audiencias:
 el `frozenset` hace que el orden de las categorías sea irrelevante ($AB = BA$).

---

````
# Listas indexadas por franja -> acceso O(1)
FRANJAS = ["V20", "S12", "S16", "S18", "S20", "D12", "D16", "D18", "D20", "L20"]

COEF = [0.40,                      # V20
        0.55, 0.70, 0.80, 1.00,    # S12 S16 S18 S20
        0.45, 0.75, 0.85, 1.00,    # D12 D16 D18 D20
        0.40]                      # L20

NUM_FRANJAS = len(FRANJAS)     # F = 10
IDX_VIERNES = 0                # V20  -> exactamente 1 partido
IDX_LUNES   = 9                # L20  -> exactamente 1 partido

# Penalizacion segun el n de COINCIDENCIAS (= partidos en la franja - 1)
PEN = [0.00, 0.25, 0.45, 0.60, 0.70, 0.75, 0.78, 0.80, 0.80]

# Tabla de audiencia base: el frozenset ignora el orden de las categorias (AB = BA).
# C-C = 0.47 se deduce de la jornada de ejemplo (la tabla del enunciado no la incluye).
BASE = {
    frozenset("AA"): 2.00, frozenset("AB"): 1.30, frozenset("AC"): 1.00,
    frozenset("BB"): 0.90, frozenset("BC"): 0.75, frozenset("CC"): 0.47,
}

def pen(coincidencias: int) -> float:
    """Penalizacion por coincidencias"""
    return PEN[min(coincidencias, len(PEN) - 1)]

def base_partido(cat1: str, cat2: str) -> float:
    """Audiencia base de un enfrentamiento"""
    return BASE[frozenset(cat1 + cat2)]

# Jornada de ejemplo del enunciado: (local, visitante, cat_local, cat_visitante)
JORNADA_EJEMPLO = [
    ("Celta", "Real Madrid", "B", "A"), ("Valencia", "R. Sociedad", "B", "A"),
    ("Mallorca", "Eibar", "C", "C"),    ("Athletic", "Barcelona", "B", "A"),
    ("Leganes", "Osasuna", "C", "C"),   ("Villarreal", "Granada", "B", "C"),
    ("Alaves", "Levante", "B", "B"),    ("Espanyol", "Sevilla", "B", "B"),
    ("Betis", "Valladolid", "B", "C"),  ("Atletico", "Getafe", "B", "B"),
]
NUM_PARTIDOS = len(JORNADA_EJEMPLO)   # P = 10

def bases_de(jornada):
    """Lista de audiencias base de una jornada."""
    return [base_partido(c1, c2) for _, _, c1, c2 in jornada]

print("Franjas:", FRANJAS)
print("Coefs  :", COEF)
print("Bases de la jornada de ejemplo:", bases_de(JORNADA_EJEMPLO))
````

# 4. ¿Cuál es la función objetivo?

A partir del vector de ocupación $n_s$, el peso efectivo de la franja $s$ es su coeficiente
rebajado por la penalización correspondiente a su nivel de ocupación:

$$w_s(n_s) \;=\; c_s\cdot\bigl(1-\mathrm{PEN}[\,n_s-1\,]\bigr), \qquad w_s(0)=0$$

y la función objetivo (audiencia total) es la suma, sobre las franjas, del peso
efectivo por la suma de las bases de los partidos:

$$\boxed{\;f(x) \;=\; \sum_{s=1}^{F} w_s(n_s)\cdot\!\!\sum_{i:\,x_i=s}\!\! b_i\;}$$

sujeta a $\;n_{\texttt{V20}} = 1\;$ y $\;n_{\texttt{L20}} = 1$. Su evaluación cuesta $O(P+F)$.

---


````
def pesos_efectivos(ocupacion):
    """w_s = coef_s * (1 - PEN[n_s - 1]) para cada franja ocupada (0 si esta vacia)."""
    return [COEF[s] * (1.0 - pen(n - 1)) if n > 0 else 0.0
            for s, n in enumerate(ocupacion)]

def audiencia(bases, x):
    """Funcion objetivo f(x): audiencia total en millones. Coste O(P + F)."""
    ocupacion = [0] * NUM_FRANJAS
    for s in x:
        ocupacion[s] += 1
    w = pesos_efectivos(ocupacion)
    return sum(bases[i] * w[x[i]] for i in range(len(x)))

def es_factible(x, idx_v=IDX_VIERNES, idx_l=IDX_LUNES):
    """Restriccion: EXACTAMENTE un partido el viernes y EXACTAMENTE uno el lunes."""
    return list(x).count(idx_v) == 1 and list(x).count(idx_l) == 1

# Verificacion: Jornada de ejemplo del enunciado (total 5,88 M)
bases = bases_de(JORNADA_EJEMPLO)
x_ejemplo = [FRANJAS.index(h) for h in
             ["V20","S12","S16","S18","S20","D16","D16","D18","D20","L20"]]
print(f"Audiencia de la jornada de ejemplo = {audiencia(bases, x_ejemplo):.4f} M")
print(f"  (el enunciado indica 5,88 M )")
print(f"  Factible (1 viernes, 1 lunes)? {es_factible(x_ejemplo)}")
````

# 6. Algoritmo para resolver el problema por fuerza bruta


```
funcion fuerza_bruta(bases):
    mejor_x, mejor_valor <- None, -infinito
    para cada x en {0..F-1}^P:                    # F^P combinaciones
        si x tiene EXACTAMENTE un viernes y un lunes:
            v <- audiencia(bases, x)               # O(P + F)
            si v > mejor_valor:
                mejor_x, mejor_valor <- x, v
    devolver mejor_x, mejor_valor
```


---


````
from itertools import product

def fuerza_bruta(bases, num_franjas=NUM_FRANJAS,
                 idx_v=IDX_VIERNES, idx_l=IDX_LUNES):
    """Enumera las F^P asignaciones y devuelve la factible de mayor audiencia
     solo viable para instancias pequeñas."""
    p = len(bases)
    mejor_x, mejor_val = None, float("-inf")
    for x in product(range(num_franjas), repeat=p):
        if x.count(idx_v) == 1 and x.count(idx_l) == 1:      # factibilidad
            v = audiencia(bases, list(x))
            if v > mejor_val:
                mejor_x, mejor_val = list(x), v
    return mejor_x, mejor_val

print("fuerza_bruta() lista. Se valida sobre instancias pequeñas en la pregunta 11.")
````

# 7. Complejidad del algoritmo por fuerza bruta

Se cuenta las operaciones elementales (OE) en función del tamaño de entrada. Evaluar $f(x)$ requiere $P$ accesos para construir la ocupación, $F$
operaciones para los pesos y $P$ productos, es decir $\Theta(P+F)$ OE. Se recorren las $F^{P}$
asignaciones y cada una se evalúa en $O(P+F)$:

$$T_{\text{fb}}(P,F) \;=\; \Theta\!\bigl(F^{P}\,(P+F)\bigr)
\;=\; \Theta\!\bigl(10^{10}\cdot 20\bigr)\;\approx\; 2\times 10^{11}\ \text{operaciones}$$

$$S_{\text{fb}}(P,F) \;=\; \Theta(P+F)$$

`product` genera las tuplas y solo se almacena la mejor solución encontrada más el vector de ocupación.

# 8. Algoritmo que mejora la complejidad. Argumentación


$f$ depende únicamente de dos cosas: (1) el vector de ocupación $n$ —que fija los pesos efectivos $w_s$— y (2) de qué base se empareja con qué peso.

Fijado $n$, los pesos quedan congelados y el problema se reduce a emparejar $P$ bases con $P$ pesos (el peso $w_s$ aparece repetido $n_s$ veces en el multiconjunto). Por la desigualdad del reordenamiento, si ambas sucesiones se ordenan en el mismo sentido
($b_{(1)}\ge\dots\ge b_{(P)}$ y $w_{(1)}\ge\dots\ge w_{(P)}$):

$$\sum_{i=1}^{P} b_{(i)}\,w_{(i)} \;\ge\; \sum_{i=1}^{P} b_{(i)}\,w_{\sigma(i)}
\qquad\forall\,\sigma\in S_P$$

Es decir: el partido de mayor audiencia base va a la franja de mayor peso efectivo, y así sucesivamente. El emparejamiento óptimo se obtiene ordenando, sin ninguna búsqueda.


Como $n_{\texttt{V20}} = n_{\texttt{L20}} = 1$ están fijados, solo hay que repartir los $P-2$
partidos restantes entre las $F-2$ franjas centrales.

```
funcion exacto_ocupaciones(bases):
    ordenar bases de mayor a menor
    para cada composicion de (P-2) en (F-2) partes:      # C(P+F-5, F-3) = 6.435
        n <- composicion con n_V = 1 y n_L = 1 insertados
        w <- pesos efectivos de n
        pesos <- multiconjunto (w_s repetido n_s veces), ordenado desc.   # O(P log P)
        val <- suma( bases[i] * pesos[i] )                # desigualdad del reordenamiento
        quedarse con la mejor composicion
    reconstruir la asignacion x a partir de la mejor ocupacion
```


Mejora el algoritmo de fuerza bruta:

1. La fuerza bruta prueba las $P!$ formas de repartir los partidos entre franjas; el reordenamiento da la mejor en una única ordenación $O(P\log P)$.
2. Muchísimos vectores $x$ distintos comparten el mismo $n$ y por tanto el mismo valor.
3. **Genera solo lo factible:** en lugar de enumerar y filtrar, evitando el 85 % del espacio.

El resultado: de $F^{P}=10^{10}$ asignaciones a **6.435** ocupaciones, (se verifica contra la fuerza bruta en la pregunta 11).

---


````
from itertools import combinations

def _composiciones(total, partes):
    """Composiciones debiles de `total` en `partes` (enteros >= 0 que suman total),
    por el metodo de "barras y estrellas".
    Cantidad generada: C(total + partes - 1, partes - 1)."""
    if partes == 0:
        if total == 0:
            yield []
        return
    for barras in combinations(range(total + partes - 1), partes - 1):
        comp, anterior = [], -1
        for bpos in barras:
            comp.append(bpos - anterior - 1)
            anterior = bpos
        comp.append(total + partes - 2 - anterior)
        yield comp


def exacto_ocupaciones(bases, num_franjas=NUM_FRANJAS,
                       idx_v=IDX_VIERNES, idx_l=IDX_LUNES):
    """enumeracion de ocupaciones + reordenamiento.

    Recorre solo las ocupaciones FACTIBLES (n_V = n_L = 1) y, para cada una,
    empareja bases y pesos ordenados de mayor a menor,
    que es el emparejamiento optimo para esa ocupacion.

    Devuelve (asignacion x, valor optimo)."""
    p = len(bases)
    orden = sorted(range(p), key=lambda i: -bases[i])     # bases descendentes
    b_orden = [bases[i] for i in orden]

    # franjas centrales: todas menos viernes y lunes
    medias = [s for s in range(num_franjas) if s not in (idx_v, idx_l)]

    mejor_val, mejor_ocu = float("-inf"), None
    for comp in _composiciones(p - 2, len(medias)):        # solo factibles
        ocu = [0] * NUM_FRANJAS
        ocu[idx_v] = ocu[idx_l] = 1                        # restriccion nativa
        for s, n in zip(medias, comp):
            ocu[s] = n
        w = pesos_efectivos(ocu)
        # multiconjunto de pesos: la franja s aporta n_s copias de w_s
        pesos = sorted((w[s] for s in range(NUM_FRANJAS) for _ in range(ocu[s])),
                       reverse=True)
        val = sum(bi * wi for bi, wi in zip(b_orden, pesos))   # reordenamiento
        if val > mejor_val:
            mejor_val, mejor_ocu = val, ocu

    # reconstruir la asignacion concreta a partir de la mejor ocupacion
    w = pesos_efectivos(mejor_ocu)
    ranuras = sorted(((w[s], s) for s in range(NUM_FRANJAS)
                      for _ in range(mejor_ocu[s])), key=lambda t: -t[0])
    x = [0] * p
    for pos, i in enumerate(orden):
        x[i] = ranuras[pos][1]
    return x, mejor_val


# Aplicacion a la jornada de ejemplo del enunciado
x_opt, v_opt = exacto_ocupaciones(bases)
print(f"Optimo de la jornada de ejemplo = {v_opt:.4f} M")
print(f"Jornada propuesta en el enunciado = {audiencia(bases, x_ejemplo):.4f} M")
print(f"Mejora = +{(v_opt/audiencia(bases, x_ejemplo) - 1)*100:.1f} %")
````

# 9. Complejidad del algoritmo mejorado

El número de ocupaciones factibles es el de composiciones débiles de $P-2$ en $F-2$ partes:

$$N(P,F) \;=\; \binom{(P-2)+(F-2)-1}{(F-2)-1} \;=\; \binom{P+F-5}{F-3}
\;=\; \binom{15}{7} \;=\; 6\,435$$

Cada ocupación cuesta $O(P\log P)$ (ordenar el multiconjunto de pesos), y la ordenación inicial
de las bases es $O(P\log P)$ una sola vez:

$$T_{\text{exacto}}(P,F) \;=\; O\!\left(\binom{P+F-5}{F-3}\; P\log P\right)
\;\approx\; 6\,435\cdot 33 \;\approx\; 2\times 10^{5}\ \text{operaciones}$$

$$S_{\text{exacto}}(P,F) \;=\; O(P+F)$$

El espacio sigue siendo lineal:

### Comparativa


| Algoritmo | Tiempo | Operaciones ($P=F=10$) | Óptimo garantizado |
|---|---|---|---|

| Fuerza bruta: | $\Theta(F^{P}(P+F))$ | $\approx 2\times 10^{11}$ | Sí |
|---|---|---|---|
| **Mejorado:** | $O\!\left(\binom{P+F-5}{F-3}P\log P\right)$ | $\approx 2\times 10^{5}$ | **Sí** |


Una mejora de ~6 órdenes de magnitud ($\times 10^{6}$)

# 10. Juego de datos de entrada aleatorios

Un dato de entrada es una jornada: un emparejamiento de los 20 equipos en 10 partidos, distribuye aleatoriamente los equipos y los empareja de dos en dos.

---


````
import random

def jornada_aleatoria(semilla=None, n_A=3, n_B=11, n_C=6):
    """Genera una jornada aleatoria respetando la distribucion del enunciado)"""
    rnd = random.Random(semilla)
    categorias = ["A"] * n_A + ["B"] * n_B + ["C"] * n_C     # 20 equipos
    equipos = [f"Eq{i:02d}" for i in range(len(categorias))]
    idx = list(range(len(equipos)))
    rnd.shuffle(idx)
    jornada = []
    for k in range(0, len(idx), 2):          # emparejar de dos en dos
        i, j = idx[k], idx[k + 1]
        jornada.append((equipos[i], equipos[j], categorias[i], categorias[j]))
    return jornada

# Jornada aleatoria reproducible
jornada_rnd = jornada_aleatoria(semilla=2026)
print("JORNADA ALEATORIA GENERADA (semilla = 2026)\n")
print(f"{'Partido':<16}{'Cat':<8}{'Base':>6}")
print("-" * 30)
for loc, vis, c1, c2 in jornada_rnd:
    print(f"{loc+'-'+vis:<16}{c1+'-'+c2:<8}{base_partido(c1,c2):>6.2f}")
print("-" * 30)
print("Bases:", bases_de(jornada_rnd))
````
