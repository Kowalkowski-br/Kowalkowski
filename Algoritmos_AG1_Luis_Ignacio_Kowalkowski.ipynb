##Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vueta atrás


```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

# Práctica individual: los dos puntos más cercanos

Problema: Encontrar los dos puntos más cercanos


- Suponer en 1D, o sea, una lista de números: [3403, 4537, 9089, 9746, 7259, …. 

- Primer intento: Fuerza bruta 

- Calcular la complejidad. ¿Se puede mejorar? 

- Segundo intento. Aplicar Divide y Vencerás 

- Calcular la complejidad. ¿Se puede mejorar? 

- Extender el algoritmo a 2D: [(1122, 6175), (135, 4076), (7296, 2741)… 

- Extender el algoritmo a 3D. 

## 1. Caso 1D: fuerza bruta


```python
from math import inf

def mas_cercanos_1d_fb(L):
    """Par de números más cercanos de una lista. Fuerza bruta O(n^2).

    Devuelve (distancia_minima, (a, b)). Si len(L) < 2 devuelve (inf, None).
    """
    n = len(L)
    if n < 2:
        return inf, None

    mejor, par = inf, None
    for i in range(n):                      # O(n)
        for j in range(i + 1, n):           # O(n)  -> n(n-1)/2 comparaciones
            d = abs(L[i] - L[j])
            if d < mejor:
                mejor, par = d, (L[i], L[j])
    return mejor, par


# Ejemplo del enunciado
L = [3403, 4537, 9089, 9746, 7259, 4520, 1122, 6175]
print("Lista :", L)
print("Fuerza bruta ->", mas_cercanos_1d_fb(L))
```

## 2. Caso 1D: divide y vencerás


**¿Se puede mejorar?** En 1D, sí y no. El coste está dominado por la ordenación, y ordenar por
comparaciones tiene una cota inferior de $\Omega(n\log n)$, así que por esta vía no se mejora.

De hecho, una vez ordenada la lista la recursión es innecesaria: basta recorrer las
diferencias consecutivas (`min(A[i+1]-A[i])`).

Sin embargo, la versión recursiva es útil porque representa el esquema que sí se necesita en 2D y 3D.


```python
def _dv_1d(A, ini, fin):
    """Recursión sobre el vector YA ORDENADO A, en el intervalo [ini, fin)."""
    n = fin - ini

    # --- Casos base ---
    if n < 2:
        return inf, None
    if n == 2:
        return A[ini + 1] - A[ini], (A[ini], A[ini + 1])

    # --- Dividir ---
    m = ini + n // 2

    # --- Vencer ---
    dI, pI = _dv_1d(A, ini, m)      # mitad izquierda
    dD, pD = _dv_1d(A, m, fin)      # mitad derecha
    d, p = (dI, pI) if dI <= dD else (dD, pD)

    # --- Combinar: la unica pareja cruzada candidata es (A[m-1], A[m]) ---
    d_cruce = A[m] - A[m - 1]
    if d_cruce < d:
        return d_cruce, (A[m - 1], A[m])
    return d, p


def mas_cercanos_1d(L):
    """Par de numeros mas cercanos. Divide y venceras O(n log n)."""
    if len(L) < 2:
        return inf, None
    A = sorted(L)                   # Theta(n log n): domina el coste total
    return _dv_1d(A, 0, len(A))     # Theta(n)


print("Lista :", L)
print("Divide y venceras ->", mas_cercanos_1d(L))
print("Coincide con fuerza bruta:", mas_cercanos_1d(L)[0] == mas_cercanos_1d_fb(L)[0])
```

## 3. Extensión a 2D



```python
import math

def distancia(p, q):
    """Distancia euclidea entre dos puntos de cualquier dimension."""
    return math.sqrt(sum((a - b) ** 2 for a, b in zip(p, q)))


def par_mas_cercano_fb(P):
    """Fuerza bruta en dimension arbitraria. O(d * n^2)."""
    n = len(P)
    if n < 2:
        return inf, None
    mejor, par = inf, None
    for i in range(n):
        for j in range(i + 1, n):
            d = distancia(P[i], P[j])
            if d < mejor:
                mejor, par = d, (P[i], P[j])
    return mejor, par


def _mezcla(A, B, eje):
    """Fusion de dos listas ordenadas por 'eje'. O(|A|+|B|), como en mergesort."""
    R, i, j = [], 0, 0
    while i < len(A) and j < len(B):
        if A[i][eje] <= B[j][eje]:
            R.append(A[i]); i += 1
        else:
            R.append(B[j]); j += 1
    R.extend(A[i:]); R.extend(B[j:])
    return R


def _franja(Py, x_med, delta, eje):
    """Busca la mejor pareja 'a caballo' de la division dentro de la franja.

    Py    : puntos de la region ordenados por 'eje' (la 2a coordenada)
    x_med : coordenada x de la division
    delta : mejor distancia encontrada hasta ahora
    """
    mejor, par = delta, None
    B = [p for p in Py if abs(p[0] - x_med) < delta]   # solo la franja: O(n)

    for i in range(len(B)):
        j = i + 1
        # Solo los vecinos con diferencia en 'eje' menor que delta:
        # como maximo un numero CONSTANTE de ellos (7 en 2D)
        while j < len(B) and (B[j][eje] - B[i][eje]) < mejor:
            d = distancia(B[i], B[j])
            if d < mejor:
                mejor, par = d, (B[i], B[j])
            j += 1
    return mejor, par


def _dv_nd(Px, eje):
    """Recursion. Px viene ordenado por x.
    Devuelve (distancia, par, Px_ordenado_por_eje)."""
    n = len(Px)

    # --- Caso base: con 3 o menos puntos, fuerza bruta directa ---
    if n <= 3:
        d, par = par_mas_cercano_fb(Px)
        return d, par, sorted(Px, key=lambda p: p[eje])

    # --- Dividir por la mediana en x ---
    m = n // 2
    x_med = Px[m][0]

    # --- Vencer ---
    dI, pI, PyI = _dv_nd(Px[:m], eje)
    dD, pD, PyD = _dv_nd(Px[m:], eje)
    d, par = (dI, pI) if dI <= dD else (dD, pD)

    # --- Combinar ---
    Py = _mezcla(PyI, PyD, eje)                 # O(n): evita reordenar la franja
    dF, pF = _franja(Py, x_med, d, eje)         # O(n)
    if pF is not None and dF < d:
        d, par = dF, pF

    return d, par, Py


def par_mas_cercano(P):
    """Los dos puntos mas cercanos de un conjunto en 1D, 2D, 3D... O(n log n)."""
    P = [tuple(p) for p in P]
    if len(P) < 2:
        return inf, None
    dim = len(P[0])
    eje = 1 if dim >= 2 else 0                  # 2o eje de barrido (y); en 1D es el propio x
    d, par, _ = _dv_nd(sorted(P), eje)          # Theta(n log n) el preordenado por x
    return d, par


P2D = [(1122, 6175), (135, 4076), (7296, 2741), (1150, 6200), (9089, 9746), (3403, 4537)]
print("2D fuerza bruta     ->", par_mas_cercano_fb(P2D))
print("2D divide y venceras->", par_mas_cercano(P2D))
```

## 4. Extensión a 3D 


```python
import random

random.seed(2024)

# --- Generacion de conjuntos aleatorios ---
def puntos_aleatorios(n, dim, maximo=10000, enteros=True):
    if enteros:
        return [tuple(random.randint(0, maximo) for _ in range(dim)) for _ in range(n)]
    return [tuple(random.uniform(0, maximo) for _ in range(dim)) for _ in range(n)]


P3D = puntos_aleatorios(12, 3, 100)
print("Conjunto 3D:", P3D, "\n")
print("3D fuerza bruta     ->", par_mas_cercano_fb(P3D))
print("3D divide y venceras->", par_mas_cercano(P3D))

# La misma funcion resuelve tambien el caso 1D (puntos de una sola coordenada)
P1D = [(x,) for x in L]
print("\n1D con la funcion generica ->", par_mas_cercano(P1D))
```
